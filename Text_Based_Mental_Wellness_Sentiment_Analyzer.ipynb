```python
import nltk
import re

from nltk.sentiment import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()
```

```python
text = """
I have been feeling a little stressed because of my studies.
There are many assignments and exams, but I am trying my best
to stay positive and manage everything.
"""
```

```python
# Sentiment analysis
score = sia.polarity_scores(text)

if score['compound'] >= 0.05:
    sentiment = "Positive 😊"
elif score['compound'] <= -0.05:
    sentiment = "Negative 😞"
else:
    sentiment = "Neutral 😐"

# Word count
words = re.findall(r'\b\w+\b', text)
word_count = len(words)

# Wellness-related keywords
wellness_keywords = [
    "happy", "sad", "stress", "stressed", "anxiety",
    "worried", "tired", "lonely", "angry", "positive",
    "peaceful", "relaxed", "pressure", "study", "exams"
]

found_keywords = []

for keyword in wellness_keywords:
    if keyword.lower() in text.lower():
        found_keywords.append(keyword)

# Display result
print("=" * 60)
print("       TEXT-BASED MENTAL WELLNESS SENTIMENT ANALYZER")
print("=" * 60)

print("\nINPUT TEXT:")
print(text)

print("\n--- SENTIMENT ANALYSIS ---")
print("Positive Score :", round(score['pos'], 2))
print("Negative Score :", round(score['neg'], 2))
print("Neutral Score  :", round(score['neu'], 2))
print("Compound Score :", round(score['compound'], 2))
print("Overall Sentiment:", sentiment)

print("\n--- TEXT STATISTICS ---")
print("Word Count:", word_count)

print("\n--- WELLNESS-RELATED KEYWORDS ---")
print(found_keywords if found_keywords else "No keywords detected")

print("\n--- WELLNESS INDICATOR ---")

if sentiment.startswith("Positive"):
    print("😊 The text shows a generally positive emotional tone.")
elif sentiment.startswith("Negative"):
    print("😞 The text shows a generally negative emotional tone.")
else:
    print("😐 The text shows a generally neutral emotional tone.")

print("\nNote: This system provides only a general sentiment indication.")
print("It is not a medical or psychological diagnosis.")

print("\n" + "=" * 60)
```
